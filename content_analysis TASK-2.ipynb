```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("cleaned_netflix.csv")
```

```python
content_count = df["type"].value_counts()

print(content_count)
```

```python
plt.figure(figsize=(6,5))
sns.countplot(data=df, x="type")

plt.title("Distribution of Netflix Content Types")
plt.xlabel("Content Type")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(6,6))

plt.pie(
    content_count,
    labels=content_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Movies vs TV Shows")
plt.show()
```

```python
percentage = (content_count / len(df)) * 100

print(percentage)
```

# Summary Report

• Movies account for the majority of Netflix content.
• TV Shows represent a smaller share of the catalog.
• The analysis shows that Netflix offers significantly more Movies than TV Shows.
