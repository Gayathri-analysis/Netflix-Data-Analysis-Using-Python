```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("cleaned_netflix.csv")
```

```python
rating_count = df["rating"].value_counts()

print(rating_count)
```

```python
plt.figure(figsize=(10,6))

sns.countplot(
    data=df,
    y="rating",
    order=df["rating"].value_counts().index
)

plt.title("Netflix Content Ratings")
plt.xlabel("Count")
plt.ylabel("Rating")

plt.show()
```

```python
genre_count = df["listed_in"].value_counts()

print(genre_count.head(10))
```

```python
top10_genres = genre_count.head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top10_genres.values,
    y=top10_genres.index
)

plt.title("Top 10 Netflix Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")

plt.show()
```

```python
rating_vs_type = pd.crosstab(df["type"], df["rating"])

print(rating_vs_type)
```

```python
plt.figure(figsize=(12,6))

sns.heatmap(
    rating_vs_type,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Content Type vs Rating")

plt.show()
```

# Audience Insights
