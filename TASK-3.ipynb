```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("cleaned_netflix.csv")
```

```python
df = df.dropna(subset=["country"])
```

```python
df["country"] = df["country"].str.strip()
```

```python
country_count = df["country"].value_counts()

print(country_count.head(10))
```

```python
top10 = country_count.head(10)

print(top10)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=top10.values,
    y=top10.index
)

plt.title("Top 10 Netflix Content Producing Countries")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.show()
```

```python
ranking = top10.reset_index()
ranking.columns = ["Country", "Content Count"]

print(ranking)
```

# Business Insights

• The United States has the highest number of Netflix titles.
• India and the United Kingdom are also among the leading content-producing countries.
• A small number of countries contribute a large share of Netflix's content library.
• Understanding country-wise distribution helps identify major content markets and potential regions for expansion.
