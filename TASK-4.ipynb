```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("cleaned_netflix.csv")
```

```python
df["release_year"].head()
```

```python
yearly_content = df["release_year"].value_counts().sort_index()

print(yearly_content)
```

```python
growth = yearly_content.diff()

print(growth)
```

```python
plt.figure(figsize=(12,6))

sns.lineplot(
    x=yearly_content.index,
    y=yearly_content.values,
    marker="o"
)

plt.title("Netflix Content Releases by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```

# Business Insights

• Netflix content production increased significantly after 2015.
• The platform experienced rapid growth in content releases during recent years.
• Some years show a decline, indicating possible changes in production strategy.
• Trend analysis helps understand Netflix's historical content expansion.
